## 0. HOW TO USE THIS PLAN

This plan is intentionally larger than what any one person needs in full depth. That is by design — it is a **reference architecture for a career**, not a single course. Section 1 gives you a tiering system so you scope it correctly instead of trying to max out every phase at once.

Every phase below carries a **tier tag**:

| Tag | Meaning |
|---|---|
| 🔴 **CORE** | Required for any competent architect, regardless of specialization. Never skip. |
| 🟡 **ADVANCED** | Required for senior/staff-level and for anyone who will own production systems end-to-end. |
| 🔵 **SPECIALIZED** | Required only if your track needs it (e.g. AI infra, real-time media, platform engineering). Pick based on the track table in Section 1. |

Work phases in the numbered order within each layer — the dependency graph in Section 6 tells you why. Within a phase, do not consider it complete until it clears every gate in Section 3.

<a name="breadth"></a>
## 1. THE BREADTH PROBLEM — AND HOW THIS PLAN SOLVES IT

The single biggest risk in a plan this comprehensive is **breadth without depth**: finishing having "heard of" fifty technologies without being able to independently design a production system with any of them. This plan solves that with three mechanisms.

### 1.1 Primary + Comparison Rule

For every category with multiple named technologies, go deep on **one primary**, and understand **one comparison** only at a conceptual level (what it trades off, not hands-on mastery). Do not implement every named technology in a category — that is a catalog, not architecture skill.

| Category | Primary (implement) | Comparison (understand only) |
|---|---|---|
| Relational DB | PostgreSQL | MySQL |
| Cache | Redis | Memcached |
| Message queue | RabbitMQ | SQS |
| Event streaming | Kafka | Redpanda / Pulsar |
| Runtime | Go **or** Node.js (pick one you'll use professionally) | Python (for AI/data tooling literacy) |
| API style | REST + gRPC | GraphQL |
| Container runtime | Docker | — |
| Orchestration | Kubernetes | Nomad (conceptual) |
| Cloud | One provider, deeply (AWS **or** GCP **or** Azure) | The other two, conceptually |
| IaC | Terraform | Pulumi (conceptual) |
| Observability | OpenTelemetry + Prometheus/Grafana | Datadog-class SaaS (conceptual) |
| Search | Elasticsearch/OpenSearch | Meilisearch/Typesense (conceptual) |
| Vector store | pgvector | A dedicated vector DB (conceptual) |
| Workflow engine | Temporal | Airflow (conceptual, if data-eng adjacent) |
| AI serving | vLLM | Triton (conceptual) |

### 1.2 Track-Based Specialization Filter

Pick a track. It tells you which 🔵 SPECIALIZED phases to go deep on and which to only skim at conceptual level.

| Track | Go deep on (🔵) | Skim only |
|---|---|---|
| **Backend / Platform Architect** (default) | Phases 14, 17, 21, 22, 25–29 | 18, 24 |
| **AI Infrastructure Architect** | Phase 24 in full, Phase 17 (queueing/GPU capacity) | 18 |
| **Real-Time / Media Architect** | Phase 18 in full | 24 |
| **Data / Analytics Architect** | Phases 5–9, 7 in full | 18, 24 |
| **Security Architect** | Phase 20 in full, Phase 26 | 18, 24 |

🔴 CORE phases apply to every track without exception.

### 1.3 Sequencing Fix

The original plan's dependency graph was correct but flat (24 phases in a straight line with no grouping), which hides how much is genuinely foundational versus genuinely optional. This plan groups everything into **10 layers** (Section 6). Finish a layer before starting the next; within "Layer 9 — Architecture Mastery," phases may be studied in parallel with Layers 5–8 once Layer 2 is complete, because governance, economics, and organizational thinking apply retroactively to everything you already built.

## 2. MASTER COMPLETION MODEL

Every phase — old or new — follows this exact sequence. A phase is not complete because the reading is finished.

```
DISCOVER REQUIREMENTS
        ↓
UNDERSTAND / READ SPECIFICATION OR THEORY
        ↓
MODEL
        ↓
DESIGN
        ↓
ESTIMATE COST
        ↓
IMPLEMENT
        ↓
TEST
        ↓
LOAD TEST
        ↓
FAILURE TEST
        ↓
SECURITY TEST
        ↓
OBSERVE
        ↓
DOCUMENT (incl. ADR + debt register entry if applicable)
        ↓
GOVERNANCE / ARCHITECTURE REVIEW
        ↓
ACCEPTANCE
```

Two steps were added versus the original model: **Discover Requirements** (you cannot design what you have not scoped) and **Estimate Cost** (every implementation decision has a cost consequence that should be known before, not after, building it).

<a name="contract"></a>
## 3. UNIVERSAL ARCHITECTURE COMPLETENESS CONTRACT

Every phase must answer all of the following. This is the original 11-part contract plus four sections that close the gaps (3.12–3.15).

### 3.1 Objective
What capability is being learned, why it matters, which later phase depends on it, what production problem it solves.

### 3.2 Requirements
Functional requirements, non-functional requirements, constraints, assumptions, dependencies, acceptance criteria, **and now**: who the stakeholders are and how the requirement was elicited (see Phase 0).

### 3.3 Architecture
Components, responsibilities, ownership, interfaces, dependencies, trust boundaries, failure boundaries, scaling boundaries.

### 3.4 Data
Data in/out, state, persistence, consistency, retention, ownership, schema/versioning — **and now**: classification (public/internal/confidential/PII), lineage, and lifecycle (see Phase 9).

### 3.5 Communication
For every path: protocol, transport, port, serialization, authN, authZ, timeout, retry, idempotency, ordering, backpressure, observability.

### 3.6 Failure
Timeout, crash, restart, dependency failure, network failure, duplicate request/message, stale data, overload, resource exhaustion, partial failure, recovery.

### 3.7 Security
AuthN, authZ, confidentiality, integrity, availability, input validation, secrets, encryption, abuse, supply chain, privacy.

### 3.8 Performance
p50/p95/p99, throughput, concurrency, CPU, memory, network, storage, connection usage — **and now**: which queueing/capacity model applies (see Phase 17.7).

### 3.9 Observability
Structured logs, metrics, traces, correlation IDs, health status, dependency status, error classification.

### 3.10 Operations
Deployment, configuration, migrations, rollback, backups, restore, alerts, runbooks, incident handling.

### 3.11 Documentation
Architecture diagram, sequence diagram, state diagram, data-flow diagram, ADR, API contract, failure-mode table, test report, operational runbook.

### 3.12 Ownership & Organization *(closes Gap: Organizational Architecture)*
Which team owns this component? Is the team stream-aligned, platform, enabling, or complicated-subsystem (Team Topologies)? What is the team's cognitive load budget, and does this component fit inside it? What is the "team API" — how do other teams interact with this team without synchronous coordination?

### 3.13 Governance *(closes Gap: Architecture Governance)*
Which architecture principle(s) does this satisfy? Does it require an exception to a standard, and if so, through what process? What fitness function (automated architecture test) enforces this decision going forward? Who is the accountable reviewer?

### 3.14 Economics *(closes Gap: Architecture Economics)*
What is the cost per unit (per request / per user / per GB / per event / per AI call)? What is the fully loaded cost including operational headcount? What is the cheapest viable alternative, and why was it rejected?

### 3.15 Debt & Evolution *(closes Gaps: Technical Debt, Architecture Evolution)*
Does this decision knowingly introduce debt? If so, classify it (deliberate/prudent, deliberate/reckless, inadvertent/prudent, inadvertent/reckless — Fowler's quadrant) and register it. What is the migration path if this decision needs to be reversed in 18 months? Is there a deprecation/sunset plan for whatever this replaces?

<a name="rules"></a>
## 4. ARCHITECTURE RULES

The original 8 rules, plus 3 new rules that close remaining gaps.

**Rule 1 — No Technology Without a Reason.** Every technology choice documents problem, context, alternatives, decision, trade-offs, consequences, and a reversal plan.

**Rule 2 — Every Component Has Ownership.** What it owns, what it doesn't, who calls it, what it calls, what data it owns, how it fails, scales, is secured, is observed — **and now, who (which team) owns it**.

**Rule 3 — Every Interface Has a Contract.** Request, response, schema, validation, authN/authZ, timeout, retry, idempotency, ordering, version, compatibility, rate limit, observability, error model.

**Rule 4 — Every Retryable Mutation Needs Idempotency.** Never assume `request → failure → retry` means the first request didn't execute.

**Rule 5 — Every Dependency Needs Failure Behavior.** Timeout, retry, backoff, jitter, circuit breaker, fallback, load shedding, alert, recovery.

**Rule 6 — Security Is Continuous.** Applies to networking, APIs, databases, caches, queues, containers, cloud, CI/CD, AI, observability — every phase, not a bolt-on phase.

**Rule 7 — Observability Is Continuous.** Do not wait for the observability phase to add logs and metrics. Every implementation gets minimum observability immediately.

**Rule 8 — Every Production Change Requires Rollback.** Deployment strategy, health checks, validation, rollback trigger/procedure, migration strategy, config rollback, feature-flag fallback.

**Rule 9 — Every Decision Has a Cost Owner.** *(closes Gap: Economics)* No architecture decision is approved without an explicit cost estimate and an owner accountable for that line item. Cost review is not optional and not deferred to Phase 28 — it happens at decision time, everywhere.

**Rule 10 — Every System Has a Governance Trail.** *(closes Gap: Governance)* Every non-trivial decision produces an ADR. Every architecture principle has at least one automated fitness function that fails a build when violated, wherever that's technically feasible (dependency direction, layering rules, forbidden imports, latency budgets, etc.).

**Rule 11 — Nothing Is Permanent; Everything Has an Exit.** *(closes Gap: Evolution)* Every architecture decision states, at the time it's made, how it would be migrated away from — the expand/contract pattern, the strangler path, or the explicit "this is meant to be permanent and here's why."

# LAYER 1 — DISCOVERY & FOUNDATIONS

## PHASE 0 — REQUIREMENTS ENGINEERING & PRODUCT DISCOVERY 🔴 CORE *(NEW — closes Gap #1)*

**Prerequisite:** None. **Unlocks:** Every other phase.

### Objective
Architecture built on unexamined requirements is architecture built on guesses. This phase teaches how to discover what actually needs to be built before any component diagram is drawn.

### 0.1 Stakeholder Identification
- Primary users, secondary users, operators, regulators, paying customer vs. end user (when different)
- RACI (Responsible / Accountable / Consulted / Informed) per major decision

### 0.2 Requirements Discovery Techniques
- Stakeholder interviews, user story mapping, job-stories ("when I ___, I want to ___, so I can ___")
- User journeys / journey mapping
- Event storming (for domain discovery — feeds directly into Phase 3 DDD work)
- Competitive/market analysis for greenfield products

### 0.3 Requirement Types
- **Business requirements** — why does this exist, what business outcome does it drive
- **Functional requirements** — what the system must do
- **Non-functional requirements (NFRs)** — performance, availability, security, compliance, usability, portability, maintainability — elicited explicitly, not assumed
- **Constraints** — budget, timeline, team size, existing systems, regulatory
- **Assumptions** — documented explicitly, revisited on schedule
- **Out-of-scope** — as important to write down as in-scope

### 0.4 Prioritization Frameworks
- MoSCoW (Must/Should/Could/Won't)
- RICE (Reach, Impact, Confidence, Effort)
- Kano model (basic / performance / delight)
- Cost of Delay / WSJF (Weighted Shortest Job First)

### 0.5 Risk Identification
- Technical risk, market risk, organizational risk, compliance risk
- Risk register: probability × impact, mitigation, owner

### 0.6 Requirement Traceability
Every requirement must be traceable through the full chain defined in Section 17 (Final Zero-Gap Checklist) — business requirement → functional requirement → NFR → architecture decision → component → interface → implementation → test → deployment → monitoring → production validation → runbook. A requirement that cannot be traced is not complete.

### 0.7 Acceptance Criteria Writing
- Given/When/Then format
- Measurable NFR thresholds (not "fast" — "p95 < 200ms at 500 RPS")
- Definition of Done at the requirement level

### Required Implementation
1. Write a full requirements document (business + functional + NFR + constraints + assumptions + out-of-scope) for the system you'll build in the capstone (Phase 31)
2. Run one event-storming session (solo or with peers) to derive a first-pass domain model
3. Produce a prioritized backlog using RICE or MoSCoW
4. Produce a risk register with at least 10 entries

### Deliverables
Requirements document, risk register, prioritized backlog, traceability matrix template (reused for the rest of the plan).

### Gate
Can you state, for the system you're about to build, every stakeholder, every NFR with a number attached, every explicit assumption, and what's deliberately out of scope? If not, this phase isn't done.

## PHASE 1 — NETWORKING & COMPUTER SCIENCE FOUNDATIONS 🔴 CORE

**Prerequisite:** Phase 0. **Unlocks:** Every subsequent technical phase.

### Objective
Understand exactly how data moves from one application to another across a real network — this was the strongest section of the original plan and is preserved in full, with the identified gaps folded in directly (marked ➕).

### 1.1 Networking Models
OSI 7-layer model, TCP/IP 4-layer model, five-layer hybrid model, encapsulation/decapsulation, headers, payloads, MTU, fragmentation, protocol boundaries.

### 1.2 IP Fundamentals
IPv4, IPv6, addressing, subnet masks, **➕ CIDR notation and subnetting practice problems**, private/public addresses, loopback, multicast, broadcast, anycast, routing, default gateway, ARP, NDP, ICMP/ICMPv6, TTL/hop limit, packet forwarding, routing tables, **➕ basic BGP concepts (how routes propagate between autonomous systems — conceptual depth is enough unless on a Network/SRE track)**.

### 1.3 TCP — Complete Coverage
4-tuple identity; TCP header fields; flags (SYN/ACK/FIN/RST/PSH/URG/ECE/CWR); full state machine (CLOSED → LISTEN → SYN-SENT/SYN-RECEIVED → ESTABLISHED → FIN-WAIT-1/2, CLOSE-WAIT, CLOSING, LAST-ACK, TIME-WAIT → CLOSED); three-way handshake; SYN backlog/cookies; four-way graceful shutdown; half-close; RST behavior.

**Reliability:** sequence numbers, cumulative/duplicate ACKs, retransmission, RTO, fast retransmit, SACK, timestamps.

**Flow control:** receive buffer, advertised window, zero window, window scaling.

**Congestion control (mandatory):** cwnd, slow start, congestion avoidance, fast retransmit/recovery, Reno/New Reno, CUBIC, BBR, ECN, congestion collapse. Understand explicitly: *flow control protects the receiver; congestion control protects the network.*

**Performance:** MSS/MTU/PMTUD, Nagle, delayed ACK, socket buffers, connection reuse, keepalive, head-of-line blocking, bandwidth-delay product (`BDP = bandwidth × RTT`).

**➕ Kernel/OS layer (closes the "runtime foundations" gap at the networking boundary):** ephemeral ports, connection tracking, syscall overhead for socket I/O, epoll/kqueue event notification, network namespaces (Linux), basic eBPF-based packet observability as a modern alternative to `tcpdump` for high-volume tracing.

### 1.4 UDP — Complete Coverage
Connectionless datagrams, message boundaries, no handshake/ordering/retransmission/delivery guarantee/congestion control/flow control at the UDP layer itself. Header: source/dest port, length, checksum. Use cases: DNS, DHCP, NTP, VoIP, RTP, WebRTC, QUIC, gaming, telemetry, service discovery, streaming. Failure modes: loss, duplication, reordering, corruption, fragmentation, NAT timeout, firewall blocking, amplification abuse.

**Application-level reliability over UDP (required implementation):** build a minimal reliable-UDP protocol with sequence numbers, ACKs, timeout, retransmission, duplicate detection, ordering, checksum/integrity, and a basic congestion/pacing concept — to internalize why reliable transport is hard.

### 1.5 TCP vs UDP
| Property | TCP | UDP |
|---|---|---|
| Connection | Yes | No |
| Reliable | Yes | No |
| Ordered | Yes | No |
| Byte stream | Yes | No (message boundaries) |
| Flow/congestion control | Yes | No (app-defined) |
| Handshake | Yes | No |
| Typical use | APIs, DB, HTTP/1.1/2 | DNS, RTP, WebRTC, QUIC |

### 1.6 Socket Programming
Implement TCP server/client (concurrent, graceful shutdown, timeout, keepalive) and UDP server/client (timeout, loss simulation, sequence numbers, ACKs) using raw sockets in at least one language.

### 1.7 DNS
Stub/recursive resolvers, root, TLD, authoritative servers, record types (A/AAAA/CNAME/MX/TXT/NS/SRV/PTR), TTL, negative caching, DNSSEC, DoH, DoT.

### 1.8 HTTP
HTTP/1.1, HTTP/2, HTTP/3 (over QUIC/UDP), methods, status codes, headers, caching, cookies, compression, content negotiation, connection reuse, multiplexing, stream management, request cancellation, retries.

### 1.9 TLS
TLS 1.2/1.3, certificate chains, CA trust, key exchange (ECDHE), forward secrecy, certificate/hostname validation, session resumption, PSK, 0-RTT, certificate rotation.

### 1.10 QUIC
Runs over UDP, integrates TLS, streams, connection IDs, loss detection, congestion control, flow control, connection migration, path validation, 0-RTT, reduced head-of-line blocking vs. TCP-based multiplexing.

### 1.11 NAT & Firewalls
SNAT/DNAT/PAT, stateful/stateless firewalls, connection tracking, NAT mapping behavior for TCP and UDP, hairpin NAT, CGNAT, NAT traversal (STUN/TURN/ICE — direct prerequisite for Phase 18).

### 1.12 Proxies
Forward, reverse, transparent, gateway, API gateway, L4 vs L7.

### 1.13 Serialization
JSON, Protobuf, Avro, Thrift, MessagePack, FlatBuffers — compared on size, CPU cost, schema evolution, compatibility, human readability, validation, versioning.

### Must Read
RFC 9293 (TCP), RFC 768/9868 (UDP), RFC 9000 (QUIC), RFC 9110 (HTTP Semantics), current TLS 1.3 RFC, RFC 8200 (IPv6), Kurose & Ross — *Computer Networking: A Top-Down Approach*.

### Tools
`curl`, `dig`, `nslookup`, `ss`, `ip`, `ping`, `traceroute`, `mtr`, `nc`, `socat`, `openssl`, `tcpdump`, `Wireshark`, `nmap`, `iperf3`.

### Required Implementation
TCP echo server, UDP echo server, reliable-UDP experiment, HTTP client/server, TLS client/server, DNS investigation, NAT experiment, packet capture investigation.

### Deliverables
OSI/TCP-IP diagram, TCP state machine, UDP flow diagram, packet anatomy diagrams, DNS resolution sequence, TLS sequence, QUIC sequence, NAT traversal diagram, full request path diagram.

### Gate
Explain end-to-end: `Browser → DNS → IP routing → TCP or UDP → TLS/QUIC → HTTP → Proxy/LB → Application`, and what happens when each layer fails.

### 12.3 Webhook Architecture *(new — this was entirely absent from the original plan)*
```
Your System → Event Occurs → Webhook Dispatcher → Delivery Attempt
     → Consumer Endpoint → 2xx? → Done
                          → Non-2xx/timeout? → Retry Queue → Backoff → DLQ
```
- **Delivery guarantees**: at-least-once by default; document this explicitly to consumers

- **Versioning**: webhook payload schema versioned independently from the REST API
- **Consumer-side failure handling**: what you do when a consumer endpoint is down for hours (backoff ceiling, eventual DLQ, manual redelivery tooling)

## PHASE 17 — SCALABILITY & PERFORMANCE ENGINEERING 🟡 ADVANCED *(EXPANDED — closes Gap #10)*

**Prerequisite:** Phase 4, Phase 10, Phase 16. **Unlocks:** Phase 21, Phase 22.

### 17.1 Scaling Mechanisms
Vertical vs. horizontal scaling, sharding, L4 vs. L7 load balancing, consistent hashing, session affinity, GeoDNS, Anycast, hot partitions/hot keys, connection ceilings, auto-scaling.

### 17.2 Scaling Loop
```
Traffic → Current Bottleneck → Scaling Trigger → Scaling Mechanism
        → New Capacity → Next Bottleneck
```

### 17.3 Capacity Planning Inputs
RPS, concurrency, bandwidth, payload size, storage growth, DB/cache/queue capacity, connections, CPU, memory.

### 17.4 Formal Performance Theory *(new — this entire subsection closes the flagged gap)*
- **Little's Law**: `L = λW` — applied beyond connection pools (Phase 10) to full-system capacity planning: queue depth, worker count, expected wait time.
- **Queueing theory basics**: M/M/1 and M/M/c queue models; utilization (`ρ = λ/μ`) and why latency explodes non-linearly as utilization approaches 1; why "average utilization looks fine" dashboards hide queueing collapse.
- **Amdahl's Law**: theoretical speedup limit from parallelizing part of a workload; why adding servers has diminishing returns when a serial bottleneck exists.
- **Universal Scalability Law (Gunther)**: extends Amdahl's Law with a *coherency/contention* penalty term — models why some systems get *slower* past a certain scale-out point, not just plateau.
- **Workload modeling**: open vs. closed workload models, realistic traffic distributions (bursty vs. steady, diurnal patterns), synthetic vs. replayed production traffic.

# LAYER 9 — ARCHITECTURE MASTERY

## PHASE 25 — ORGANIZATIONAL ARCHITECTURE 🟡 ADVANCED *(NEW — closes Gap #3)*

**Prerequisite:** Phase 3 (start alongside Layer 5). **Unlocks:** Phase 26, Phase 30.

### Objective
Software architecture and organizational structure are not independent — this phase makes that relationship an explicit design tool instead of an unexamined constraint.

### 25.1 Conway's Law
*"Organizations design systems that mirror their own communication structure."* Recognize this as a predictive law, not just a warning: your service boundaries will converge on your team boundaries whether you plan for it or not.

### 25.2 The Reverse Conway Maneuver
Deliberately restructure teams to produce the desired architecture, instead of letting architecture passively emerge from an accidental org chart.

### 25.3 Team Topologies
Four fundamental team types:
- **Stream-aligned team** — owns a full slice of business value end to end
- **Platform team** — provides self-service capability to stream-aligned teams, reducing their cognitive load
- **Enabling team** — temporarily helps a stream-aligned team acquire a capability, then leaves
- **Complicated-subsystem team** — owns a component requiring deep specialist knowledge (e.g. a video codec, a matching engine)

Three interaction modes: **collaboration** (close, temporary, high-bandwidth), **X-as-a-Service** (clean API, low-bandwidth), **facilitating** (one team coaches another).

### 25.4 Cognitive Load
Every team has a finite cognitive-load budget. Map it: intrinsic load (the problem domain itself), extraneous load (tooling/process friction), germane load (learning specific to the task). A service boundary that pushes a team over its cognitive-load budget is a mis-drawn boundary, regardless of how clean the code looks.

### 25.5 The Team API
Every team should be able to state, explicitly: what we own, how to interact with us (docs, on-call, SLAs), what we don't own, and what we need from other teams to succeed. This is Phase 3.6/Rule 2's "ownership" made organizational.

### 25.6 Platform Engineering & Developer Experience
Platform-as-a-product mindset: the platform team's customers are internal engineers; the platform's job is to absorb complexity so stream-aligned teams don't have to hold it. DevEx as a measurable architecture output (deployment lead time, time-to-first-commit, self-service capability coverage).

### Required Implementation
Map your capstone system's service boundaries (from Phase 14) against a proposed team structure using Team Topologies. Identify at least one place where the "natural" service boundary and a sane team boundary conflict, and decide — explicitly, in writing — which one yields.

## PHASE 26 — ARCHITECTURE GOVERNANCE & FITNESS FUNCTIONS 🟡 ADVANCED *(NEW — closes Gap #4)*

**Prerequisite:** Phase 3, Phase 25. **Unlocks:** Phase 27, Phase 29.

### Objective
ADRs alone are necessary but not sufficient governance. This phase builds the structures that keep architecture decisions enforced over time, not just documented once.

### 26.1 Architecture Principles
A short, durable set of statements the organization holds itself to (e.g. "services own their data; no shared-database access," "every public API is versioned"). Principles should be few, specific, and testable — not aspirational slogans.

### 26.2 Standards vs. Principles
Principles are why; standards are how (e.g., "all services log in structured JSON with a `request_id` field" is a standard derived from an observability principle).

### 26.3 Exception Process
No standard survives contact with reality without exceptions. Define: who can grant an exception, what has to be documented (why the standard doesn't fit, what the compensating control is, when it will be revisited), and where exceptions are tracked so they don't silently become permanent.

### 26.4 Architecture Review Board (ARB) — right-sized
For small orgs this can be one senior architect plus a lightweight async review; for larger orgs, a real board. Either way, define: what triggers a review (new external dependency, new datastore, cross-team API, a principle exception request), what the review checks against (the completeness contract in Section 3), and the maximum acceptable review latency (a governance process that blocks shipping for weeks is a governance failure, not a governance success).

### 26.5 Fitness Functions (deepened from Phase 3.5)
Categorize and implement: atomic (single characteristic, e.g. "no cyclic package dependencies") vs. holistic (combination, e.g. "p99 latency stays under budget as fan-out increases"); triggered (run in CI) vs. continual (run constantly in production, e.g. an SLO burn-rate check); static vs. dynamic threshold. Wire at least one fitness function into the CI pipeline built in Phase 23 so a build fails when an architecture principle is violated.

### 26.6 Dependency Governance
Explicit rules for what a service is allowed to depend on (internal libraries, external vendors, other teams' services); a lightweight approval path for adding a new external dependency, cross-referenced against Phase 20's supply-chain security requirements.

### 26.7 Lifecycle & Deprecation Governance
Every architecture component and API gets an explicit lifecycle state (proposed → active → deprecated → sunset → removed) with governance-tracked transitions, feeding directly into Phase 29's evolution work.

### 26.8 Architecture Maturity Model
A simple self-assessment ladder (e.g., Ad hoc → Documented → Governed → Fitness-function-enforced → Continuously validated) to know where your organization actually sits, honestly, rather than aspirationally.

### Required Implementation
Write your organization's (or capstone project's) architecture principles (5–8 max). Define one exception process. Implement two fitness functions in CI — one structural (e.g., dependency direction) and one behavioral (e.g., a latency budget test).

## PHASE 27 — TECHNICAL DEBT MANAGEMENT 🟡 ADVANCED *(NEW — closes Gap #5)*

**Prerequisite:** Phase 3, Phase 26. **Unlocks:** Phase 29.

### Objective
Every real system accumulates debt. The skill is not avoiding it entirely — it's identifying it, pricing it, and repaying it deliberately instead of by accident.

### 27.1 The Technical Debt Quadrant (Fowler)
| | Reckless | Prudent |
|---|---|---|
| **Deliberate** | "We don't have time to design" | "We must ship now and deal with consequences" |
| **Inadvertent** | "What's layering?" | "Now we know how we should have done it" |

Every debt entry gets classified into one of these four quadrants — it changes how urgently and how it should be repaid.

### 27.2 Debt Taxonomy Beyond Code
- **Architectural debt** — a boundary or dependency that no longer fits the domain
- **Operational debt** — manual steps that should be automated, missing runbooks
- **Security debt** — deferred patching, missing controls (cross-reference Phase 20)
- **Dependency debt** — outdated libraries/frameworks, unpatched CVEs
- **Data debt** — undocumented schemas, missing lineage, orphaned datasets (cross-reference Phase 9)
- **Test debt** — missing or flaky test coverage that slows every future change

### 27.3 The Debt Register
A tracked, visible list — not tribal knowledge. Each entry: description, quadrant classification, estimated "interest rate" (how much slower does every related change get because of this debt?), estimated repayment cost, owner, and a repay/accept-permanently decision with a date to revisit.

### 27.4 Quantification
Debt is worth pricing the way financial debt is: **principal** (cost to fix now) vs. **interest** (recurring cost of not fixing — slower delivery, more incidents, more onboarding friction). A debt item with high interest and low principal is always worth repaying immediately; the harder calls are high-principal, low-interest items, which are legitimate to defer indefinitely.

### 27.5 Prioritization & Repayment Strategy
Debt repayment competes with feature work for the same capacity — make that trade-off explicit rather than hidden: a fixed percentage of every sprint/cycle reserved for debt repayment; "boy scout rule" incremental repayment during adjacent feature work; and dedicated debt-focused cycles for large structural items that can't be chipped away incrementally.

### 27.6 Debt vs. Velocity
Track delivery lead time over time as a leading indicator of unmanaged debt — rising lead time with flat scope is a strong signal that debt interest is compounding.

### Required Implementation
Create a debt register for your capstone system with at least 10 real entries, each classified by quadrant with an interest/principal estimate. Repay at least one item and record the measurable before/after change (e.g., build time, deployment lead time, defect rate for that area).

## PHASE 28 — ARCHITECTURE ECONOMICS & FINOPS 🟡 ADVANCED *(NEW — closes Gap #2)*

**Prerequisite:** Phase 22, Phase 26. **Unlocks:** Phase 29, Phase 30.

### Objective
"What is the cost of this decision?" should be asked as often as "does this scale?" — this phase makes cost a first-class, continuously-tracked architectural property instead of a once-a-year exercise.

### 28.1 Unit Economics
Track and know, per system: cost per request, cost per user (often segmented by active vs. total), cost per GB stored/transferred, cost per event processed, cost per AI request/token. These numbers should be attached to the architecture diagram, not buried in a separate finance spreadsheet.

### 28.2 Total Cost of Ownership (TCO)
Beyond infrastructure spend: engineering time to build, engineering time to operate, on-call/incident cost, opportunity cost of the team not building something else, licensing, and the compounding cost of technical debt (Phase 27) associated with the decision.

### 28.3 Build vs. Buy
| Dimension | Question |
|---|---|
| Cost | What is TCO? |
| Time | How quickly can it ship? |
| Reliability | Who operates it? |
| Security | Who controls the data? |
| Scale | What limits exist? |
| Control | What customization is required? |
| Maintenance | Who owns upgrades? |
| Lock-in | How hard is migration away? |
| Expertise | Do we have the skill in-house? |
| Exit | Can we replace it later? |

### 28.4 Cost Attribution
Chargeback (billing internal teams for their actual usage) vs. showback (visibility without billing) — pick deliberately based on organizational maturity; tag every cloud resource with an owning team and cost center as a non-negotiable deployment gate.

### 28.5 Cost Governance
Budget alerts and anomaly detection wired into the observability stack from Phase 19 (a cost spike is an incident, not just a finance surprise); a lightweight approval gate for any change with a material cost delta, mirroring the exception process in Phase 26.

### 28.6 Cloud Cost Optimization Techniques
Reserved/committed-use discounts vs. on-demand vs. spot/preemptible instances; right-sizing based on actual utilization data (not initial guesses); autoscaling tuned for cost as well as performance; storage tiering (hot/warm/cold, cross-reference Phase 9.4); egress cost awareness in multi-region/multi-cloud designs (Phase 22).

### 28.7 FinOps Maturity Model
**Inform** (visibility into cost exists) → **Optimize** (teams act on that visibility) → **Operate** (cost efficiency is a continuous, automated practice, not a periodic cleanup project).

### Required Implementation
Instrument your capstone system to report cost-per-request (even as an estimate derived from cloud pricing). Complete one real build-vs-buy analysis for a component in your capstone using the table in 28.3. Set one cost anomaly alert tied to your observability stack.

## PHASE 29 — ARCHITECTURE EVOLUTION & MIGRATION STRATEGY 🟡 ADVANCED *(NEW — closes Gap #9)*

**Prerequisite:** Phase 26, Phase 27. **Unlocks:** Phase 30, Phase 31.

### Objective
Architecture is never finished — this phase teaches how to change a running production system safely, which is a distinct skill from designing one from scratch.

### 29.1 Evolutionary Architecture Principles
Incremental, guided change as a designed-in property (built on Phase 3.5's fitness functions); the goal is not to predict the future correctly but to make future change cheap regardless of which future arrives.

### 29.2 Migration Patterns
- **Strangler Fig** — route an increasing slice of traffic to the new implementation while the old one still exists, until it can be removed
- **Parallel Run** — run old and new systems simultaneously on live traffic, compare outputs, without the new system's output being authoritative yet
- **Shadow Traffic** — mirror production traffic to a new system for testing without affecting real responses
- **Dual Writes** — write to both old and new stores during a data migration; requires careful handling of write-order and partial-failure (an outbox-style approach, cross-reference Phase 15, is safer than naive dual writes)
- **Expand/Contract (Parallel Change)** — expand the schema/interface to support both old and new consumers, migrate consumers one at a time, then contract by removing the old path once nothing depends on it
- **Backfills** — populating historical data into a new store/schema, done in a way that doesn't starve live traffic of resources

### 29.3 Version Migration & Deprecation
Ties directly to Phase 12.2 and Phase 26.7: announce deprecation with a concrete sunset date, provide a migration guide, monitor consumer migration progress, and only remove the old path once usage is at (or accepted to be at) zero.

### 29.4 Feature Flags As An Evolution Tool
Not just for release control — use flags to gate a migration path per-tenant or per-percentage, enabling gradual rollout and instant rollback of an architecture change without a redeploy.

### 29.5 Compatibility Testing
Automated backward/forward compatibility tests as a required CI gate (implemented already for APIs in Phase 12 and schemas in Phase 9.6) — extend the same discipline to internal library and infrastructure changes.

### 29.6 Migration Runbooks
Every non-trivial migration gets a runbook: pre-migration checklist, execution steps, verification steps, rollback trigger and procedure, and a named owner during the migration window.

### Required Implementation
Design and execute one real migration on your capstone system using the expand/contract pattern (e.g., migrating a field, a table, or a service boundary). Document it as a full migration runbook, including the rollback path, and prove the rollback actually works by triggering it deliberately in a non-production environment.
